# Observation Application

Import CIPO to retrieve and review today's MPC ephemerides.


```python
import cipo
```

### Ephemerides for today's UTC date


The workflow uses the current UTC date automatically. If the MPC does not
provide ephemerides for today, it does not substitute another date.


```python
filtered_objects = cipo.process_mpc_data(
    interactive_mode=False,
    print_visible=False,
)
```

```python
if "filtered_objects" not in globals():
    print("Run the processing cell above first.")
elif filtered_objects.empty:
    print("No objects meet today's (UTC) visibility criteria.")
else:
    display(filtered_objects)
```

```python
if "filtered_objects" not in globals():
    print("Run the processing cell above first.")
elif filtered_objects.empty:
    print("There are no objects to inspect for this date.")
else:
    designation = input("Enter an object's designation to inspect its ephemerides: ").strip()
    object_ephemerides = filtered_objects[
        filtered_objects["Temp Desig"] == designation
    ]

    if object_ephemerides.empty:
        print(f"No ephemerides found for {designation} in the visible-object list.")
    else:
        print(f"Ephemerides for {designation}:")
        display(object_ephemerides)
```
