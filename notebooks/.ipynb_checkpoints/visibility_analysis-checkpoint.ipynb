```python
import pandas as pd
import cipo

OBSERVATORY_CODE = "Y28"
OBJECT_PAGE = "NEOCP"
MINIMUM_ALTITUDE_DEGREES = 20
MINIMUM_DURATION_MINUTES = 45
PLOT_ALL_OBJECTS = False  # True plots every parsed object; False plots only selected candidates.
SHOW_OBJECT_NAMES = False
```

```python
analysis_date = pd.Timestamp.now(tz="UTC").date()

result_dataframe = cipo.analyze_ephemeris_objects(
    obs_code=OBSERVATORY_CODE,
    obj_type=OBJECT_PAGE,
    altitude_min=MINIMUM_ALTITUDE_DEGREES,
    duration_min=MINIMUM_DURATION_MINUTES,
    reference_date=analysis_date,
    plot=True,
    plot_all=PLOT_ALL_OBJECTS,
    show_object_names=SHOW_OBJECT_NAMES,
    print_visible=False,
)

display(result_dataframe[['Temp Desig', 'R.A.', 'Decl.', 'V', 'Visible_Minutes', 'Max_Alt']])
```
