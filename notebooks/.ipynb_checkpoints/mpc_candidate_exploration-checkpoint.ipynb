# MPC Candidate Exploration

Retrieve current NEOCP or PCCP candidates for OASI and inspect the ranked list for today's UTC date. The results are provisional MPC candidates, not confirmed orbits.


```python
import cipo

OBSERVATORY_CODE = "Y28"
OBJECT_PAGE = "NEOCP"  # Set to "PCCP" for possible comet candidates.
MAX_MAGNITUDE = 18.0
MINIMUM_ALTITUDE_DEGREES = 10.0
MINIMUM_DURATION_MINUTES = 30
```

```python
ranked_candidates = cipo.get_prioritized_observable_objects(
    obs_code=OBSERVATORY_CODE,
    obj_type=OBJECT_PAGE,
    magnitude_max=MAX_MAGNITUDE,
    altitude_min=MINIMUM_ALTITUDE_DEGREES,
    duration_min=MINIMUM_DURATION_MINUTES,
)
```

```python
display(ranked_candidates)
```
